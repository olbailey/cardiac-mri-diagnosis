# Analysing Raw MRI Data

```python
import numpy as np
import os

from data.stats import patient, load_raw_data, process_patient

DATA_DIR = "../data/raw"

patients = load_raw_data(DATA_DIR)
print(f"Number of patients: {len(patients)}")

for i in range(5):
    print(patients[i]) 
```

## Slice Image sizes

```python
image_sizes = []
for person in patients:
    image_sizes.append(person.resized_volume.GetSize()[2])

arr = np.array(image_sizes)

print("Statistics for Image slice sizes")

print(f"sum: {np.sum(arr)}")
print()
print("Shapes (value, slice index)")
print(f"max: {int(np.max(arr, axis=0)), int(np.argmax(arr, axis=0))}")
print(f"min: {int(np.min(arr, axis=0)), int(np.argmin(arr, axis=0))}")
print()
print(f"median: {np.median(arr, axis=0)}")
print(f"mean: {np.mean(arr, axis=0)}")
print(f"std: {np.std(arr, axis=0)}")
```

```python
voxel_sizes = []
for person in patients:
    voxel_sizes.append(person.voxel_size)

arr = np.array(voxel_sizes)

print("Statistics for MRI voxal sizes")

print("Shapes (value, slice index)")
print(f"max: {float(np.max(arr, axis=0)[0]), int(np.argmax(arr, axis=0)[0])}")
print(f"min: {float(np.min(arr, axis=0)[0]), int(np.argmin(arr, axis=0)[0])}")
print(f"median: {np.median(arr, axis=0)}")
print(f"mean: {np.mean(arr, axis=0)[0]}")
print(f"std: {np.std(arr, axis=0)[0]}")
```
