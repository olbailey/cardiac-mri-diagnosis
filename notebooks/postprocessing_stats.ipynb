```python
import numpy as np
import torch

from utils.graphs import show_slice_with_mask
from utils.postprocessing import load_postprocessed_data

slice_pairs, sorted_entries = load_postprocessed_data("../")
```

```python
image, segmentation_mask = torch.load("../data/postprocessed/training/patient001_slice08_DCM.pt")
show_slice_with_mask(image, segmentation_mask, "Segmented Heart slice")
```

```python
sizes = []

for image, mask in slice_pairs:
    sizes.append(image.shape)

arr = np.array(sizes)


print(len(slice_pairs))

print(f"max: {np.max(arr, axis=0), np.argmax(arr, axis=0)}")
print(f"min: {np.min(arr, axis=0), np.argmin(arr, axis=0)}")
print()
print(f"median: {np.median(arr, axis=0)}")
print(f"mean: {np.mean(arr, axis=0)}")
print(f"std: {np.std(arr, axis=0)}")
```

```python
aspect_ratios = arr[:, 0] / arr[:, 1]

print("Aspect Ratios\n")

print(f"max: {np.max(aspect_ratios, axis=0), np.argmax(aspect_ratios, axis=0)}")
print(f"min: {np.min(aspect_ratios, axis=0), np.argmin(aspect_ratios, axis=0)}")
print(f"median: {np.median(aspect_ratios, axis=0)}")
print(f"mean: {np.mean(aspect_ratios, axis=0)}")
print(f"std: {np.std(aspect_ratios, axis=0)}")
```
